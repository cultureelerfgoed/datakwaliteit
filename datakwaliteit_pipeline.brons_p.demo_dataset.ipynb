### Kolom `aantal_1`

```python
# kolomwaarde is tussen
dq02_kolomwaarde_is_tussen(
    laag_1 = 'brons_p', 
    schema_1 = 'demo_dataset', 
    tabel_1 = 'nederland', 
    sleutelkolom_1 = 'id',
    kolom_1 = 'aantal_1',  #datatype = getal
    kleiner_dan = 100,
    groter_dan = 5
    )
```

### Kolom `gemeente_email`

```python
dq03_kolomwaarde_heeft_patroon(
    laag_1 = 'brons_p', 
    schema_1 = 'demo_dataset', 
    tabel_1 = 'nederland', 
    sleutelkolom_1 = 'id',
    kolom_1 = 'gemeente_email', #datatype = tekst
    patroon = r"^[\w-\.]+@([\w-]+\.)+[\w-]{2,4}$" #email
    )
```

### Kolom `gemeente_naam`

```python
# kolomwaarde begint niet met
dq04_kolomwaarde_begint_niet_met(
    laag_1 = 'brons_p', 
    schema_1 = 'demo_dataset', 
    tabel_1 = 'nederland', 
    sleutelkolom_1 = 'id',
    kolom_1 = 'gemeente_naam', #datatype = tekst
    patroon = "^[.]|^[ ]" #begint met spatie of punt
    )
```

### Kolom `beschrijving`

```python
# kolomwaarde eindigt niet op
dq05_kolomwaarde_eindigt_niet_op(
    laag_1 = 'brons_p', 
    schema_1 = 'demo_dataset', 
    tabel_1 = 'nederland', 
    sleutelkolom_1 = 'id',
    kolom_1 = 'beschrijving', #datatype = tekst
    patroon = "[?]$|[.]$" #eindigd op vraagteken of punt
    )

# kolomwaarde heeft lengte
dq06_kolomwaarde_heeft_lengte(
    laag_1 = 'brons_p', 
    schema_1 = 'demo_dataset', 
    tabel_1 = 'nederland', 
    sleutelkolom_1 = 'id',
    kolom_1 = 'beschrijving',  #datatype = tekst
    korter_dan = 50,
    langer_dan = 10
    )
```
