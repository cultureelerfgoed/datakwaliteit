## Start de DuckLake extensie

```python
#%pip install nbformat
import nbformat
import duckdb
from datetime import datetime
con = duckdb.connect(":memory:", config={"allow_unsigned_extensions": "true"})
#con.execute("SET extensions_directory = './duckdb/extensions'")
print(f"{datetime.now().strftime('%H:%M:%S')}\tDuckLake wordt geladen")
con.execute("INSTALL ducklake;")
con.execute("LOAD ducklake;")
exttensie_is_geladen = con.sql("""SELECT count(*)
FROM duckdb_extensions() 
WHERE extension_name = 'ducklake'
AND installed = true
AND loaded = true;""").fetchall()
if exttensie_is_geladen != 0:
    print(f"{datetime.now().strftime('%H:%M:%S')}\tDuckLake is succesvol geladen")
else:
    print(f"{datetime.now().strftime('%H:%M:%S')}\tDuckLake is NIET geladen")
```
