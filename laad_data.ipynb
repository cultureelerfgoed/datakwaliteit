## Maak de catalogi
Voor elke medallion architectuur laag wordt een catalogus aangemaakt
- brons
- zilver
- goud

Voor het toepassen van OTAP kunnen de volgende postfixes toegevoegd worden (optioneel)
- _p = productie
- _t = test
- _a = accesptatie
- _o = ontwikkel

```python
con.execute("ATTACH IF NOT EXISTS 'ducklake:brons_p.ducklake' AS brons_p;")
con.execute("ATTACH IF NOT EXISTS 'ducklake:zilver_p.ducklake' AS zilver_p;")
con.execute("ATTACH IF NOT EXISTS 'ducklake:goud_p.ducklake' AS goud_p;")
print(f"{datetime.now().strftime('%H:%M:%S')}\tCatalogi zijn aangemaakt")
```

## Maak schema's

```python
con.execute("CREATE SCHEMA IF NOT EXISTS brons_p.demo_dataset;")
con.execute("CREATE SCHEMA IF NOT EXISTS brons_p.referentiedata;")
con.execute("CREATE SCHEMA IF NOT EXISTS brons_p.datakwaliteit;")
print(f"{datetime.now().strftime('%H:%M:%S')}\tSchema's zijn aangemaakt")
```

## Maak tabellen


```python
con.execute("CREATE TABLE IF NOT EXISTS brons_p.demo_dataset.nederland AS SELECT * FROM './brondata/dataset_nederland.csv';")
con.execute("CREATE TABLE IF NOT EXISTS brons_p.referentiedata.gemeente_lijst AS SELECT * FROM './brondata/referentiedata_gemeente.csv';")
con.execute("CREATE TABLE IF NOT EXISTS brons_p.referentiedata.provincie_lijst AS SELECT * FROM './brondata/referentiedata_provincie.csv';")
con.execute("CREATE TABLE IF NOT EXISTS brons_p.datakwaliteit.log AS SELECT * FROM './brondata/kwaliteit_log_template.csv';")
print(f"{datetime.now().strftime('%H:%M:%S')}\tTabellen zijn aangemaakt")
print(f"{datetime.now().strftime('%H:%M:%S')}\tData zijn in de tabellen geladen")
```
