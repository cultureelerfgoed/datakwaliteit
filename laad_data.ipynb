## Maak de catalogi
Voor elke medallion architectuur laag wordt een catalogus aangemaakt
- brons
- zilver
- goud

Voor het toepassen van OTAP kunnen de volgende postfixes toegevoegd worden (optioneel)
- _p = productie
- _t = test
- _a = accesptatie
- _o = ontwikkel

```python
con.execute("ATTACH IF NOT EXISTS 'ducklake:brons_p.ducklake' AS brons_p;")
con.execute("ATTACH IF NOT EXISTS 'ducklake:zilver_p.ducklake' AS zilver_p;")
con.execute("ATTACH IF NOT EXISTS 'ducklake:goud_p.ducklake' AS goud_p;")
print(f"{datetime.now().strftime('%H:%M:%S')}\tCatalogi zijn aangemaakt")
```

## Maak schema's

```python
con.execute("CREATE SCHEMA IF NOT EXISTS brons_p.demo_dataset;")
con.execute("CREATE SCHEMA IF NOT EXISTS brons_p.referentiedata;")
con.execute("CREATE SCHEMA IF NOT EXISTS brons_p.datakwaliteit;")
con.execute("CREATE SCHEMA IF NOT EXISTS brons_p.complexen;")
con.execute("CREATE SCHEMA IF NOT EXISTS brons_p.abr")
print(f"{datetime.now().strftime('%H:%M:%S')}\tSchema's zijn aangemaakt")
```
